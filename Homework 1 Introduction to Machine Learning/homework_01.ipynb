## Q6. Median horsepower after filling missing values with the mode

```python
median_before = df['horsepower'].median()
mode_hp = df['horsepower'].mode().iloc[0]
df['horsepower'] = df['horsepower'].fillna(mode_hp)
median_after = df['horsepower'].median()
median_before, mode_hp, median_after, median_before == median_after
```

Expected answer: **No** — the median does not change.

## Q7. Sum of weights

```python
asia = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
X = asia.to_numpy()
XTX = X.T @ X
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = np.linalg.inv(XTX) @ X.T @ y
X, w, w.sum()
```

Expected answer: **0.369** (calculated value ≈ 0.369196969).
